````python
from transformers import pipeline

# Load the chat model (same model we used in class)
pipe = pipeline('text-generation', model='TinyLlama/TinyLlama-1.1B-Chat-v1.0')

# Helper function (from class)
def get_response(messages, temperature=0.5, max_new_tokens=100):
    prompt = pipe.tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    outputs = pipe(prompt, max_new_tokens=max_new_tokens, do_sample=True, temperature=temperature)
    return outputs[0]["generated_text"].split("<|assistant|>")[-1]

print("Setup complete! ✅")
````

### Exercise 1.2: Rules of Engagement — Constraints

System prompts are not just for personas — they are for **rules** too!

**Task:** Write a system prompt that forces the AI to:
1. Answer in **a maximum of 2 sentences** (Constraint)
2. Avoid using **any complex/technical words** (Constraint)

Question: *"How does the internet work?"*


````python
# TODO: Write a system prompt with constraints
# System prompt with constraints
messages_rules = [
    {
        "role": "system",
        "content": "Answer in no more than 2 sentences. Use only simple, everyday words and avoid any complex or technical terms."
    },
    {
        "role": "user",
        "content": "How does the internet work?"
    }
]

print(get_response(messages_rules))
````

**🤔 Reflection Question 1.2:** Did the model actually follow your rules? If not, how could you improve the prompt? (Hint: small models often ignore rules — try making your prompt stricter and clearer!)

> The model followed some of the rules, but not all of them. It sometimes used words that were a little technical or gave a longer answer than requested because TinyLlama is a small model and may not always follow instructions exactly. To improve the results, I would make the prompt more specific, for example: "You must answer in exactly 2 sentences. Use only simple words that a 10-year-old can understand. Do not use any technical terms. If you cannot follow these rules, do not answer." Clearer and stricter instructions help the model follow the constraints more consistently.

### Exercise 2.2: The Repeat Test — Deterministic or Random? 🎲

**Task:** Run the same prompt **3 times at the same temperature**.
- First, 3 runs at `temperature = 0.1`
- Then, 3 runs at `temperature = 1.2`

Observe: in which case are the outputs more similar to each other?



**Observation:** The outputs at temperature = 0.1 were much more similar to each other because the model consistently selected the most likely next words. The outputs at temperature = 1.2 were more varied, with different ideas, wording, and story directions in each run due to the higher randomness in token selection.

````python
prompt = "The most important invention in history is"

# TODO: loop 3 times at temperature 0.1
for i in range(3):
    res = generator(prompt, max_length=30, do_sample=True, temperature=0.1)  # <-- fix
    print(f"Run {i+1}:", res[0]['generated_text'])
    print("-" * 60)
````

````python
# TODO: now 3 times at temperature 1.2
for i in range(3):
    res = generator(prompt, max_length=30, do_sample=True, temperature=1.2)  # <-- fix
    print(f"Run {i+1}:", res[0]['generated_text'])
    print("-" * 60)
````

**🤔 Reflection Question 2.2:** Why were the outputs similar at low temperature? Explain in terms of the probability distribution (Hint: remember the "Serious Student" analogy — it only picks the highest-probability word).

> At a low temperature (0.1), the probability distribution becomes sharper, so the highest-probability words are much more likely to be selected. As a result, the model makes nearly the same choices each time, producing very similar outputs. This is like the "Serious Student" analogy—the model sticks to the safest and most likely answer instead of exploring different possibilities.

## Part 3: Top-P (Nucleus Sampling) — The "Selection Pool" 🎯

Remember:
- **Top-P = 0.1** → the AI only looks at the top 1-2 words (probability sum = 10%)
- **Top-P = 0.9** → a bigger pool, more creativity, but higher chance of mistakes


### Exercise 3.1: Top-P Comparison

**Task:** Same prompt, different top_p values: `0.1`, `0.5`, `0.95`

Prompt: `"In a distant galaxy, humans discovered a planet made of"`


````python
prompt = "In a distant galaxy, humans discovered a planet made of"

# TODO: top_p = 0.1
res = generator(prompt, max_length=50, do_sample=True, top_p=0.1)  # <-- fix
print("TOP-P 0.1:\n", res[0]['generated_text'])
````

````python
# TODO: top_p = 0.5
res = generator(prompt, max_length=50, do_sample=True, top_p=0.5)  # <-- fix
print("TOP-P 0.5:\n", res[0]['generated_text'])
````

````python
# TODO: top_p = 0.95
res = generator(prompt, max_length=50, do_sample=True, top_p=0.95)  # <-- fix
print("TOP-P 0.95:\n", res[0]['generated_text'])
````

**🤔 Reflection Question 3.1:** Temperature and Top-P — both control randomness, but the **mechanism is different**. Explain the difference between the two in your own words:

| Feature          | Temperature                                                      | Top-p (Nucleus Sampling)                                                                                   |
| ---------------- | ---------------------------------------------------------------- | ---------------------------------------------------------------------------------------------------------- |
| **How it works** | Changes the probability distribution of all possible next words. | Limits the model to the smallest set of words whose cumulative probability reaches the chosen **p** value. |
| **Effect**       | Controls how random or predictable the output is.                | Controls how many candidate words the model can choose from.                                               |
| **Low value**    | More deterministic and repetitive.                               | Chooses from only the most likely words, making output more focused.                                       |
| **High value**   | More creative but may become less coherent.                      | Includes more possible words, increasing diversity and creativity.                                         |
| **Main idea**    | Adjusts the likelihood of every word.                            | Restricts the pool of words before sampling.                                                               |
