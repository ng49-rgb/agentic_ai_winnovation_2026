### Q1. LLM vs Agent
What is the difference between a plain LLM and an AI Agent? Give one real-life example of a task that an LLM alone **cannot** do, but an Agent **can**.

**Your answer:**

LLM (Large Language Model):

An LLM generates text based on the prompt it receives.
It relies on the knowledge it learned during training.
It cannot directly access external tools, APIs, databases, or perform real-world actions unless integrated with them.

AI Agent:

An AI Agent uses an LLM as its reasoning engine but can also interact with tools, APIs, databases, calculators, web search, and other software.
It follows a reasoning process (such as ReAct) to decide which tool to use, observe the results, and continue until it reaches the final answer.

| LLM                                   | AI Agent                                      |
| ------------------------------------- | --------------------------------------------- |
| Generates text only                   | Can reason and use tools                      |
| No memory of actions                  | Can perform multi-step tasks                  |
| Cannot interact with external systems | Can call APIs, databases, calculators, etc.   |
| Answers from training data            | Combines reasoning with real-time information |

**Your answer:**

Every tool has a docstring (description) because the LLM reads the descriptions to decide which tool to use.

The description tells the agent:

What the tool does.
When it should be used.
What type of input it expects.
What output it returns.

The agent may choose the wrong tool.

It may fail to solve the task.

### Exercise B1 — Currency Converter Tool (Easy)

Create a tool called `currency_converter` that converts USD to INR.

Requirements:
* Input: a string amount in USD, e.g. `"100"`
* Use a fixed rate: 1 USD = 83 INR
* Return a string like: `"100 USD = 8300 INR"`
* Don't forget the docstring — the agent can't use the tool without it!

```python
# tool
# def currency_converter(amount_usd: str) -> str:
#     """
#     # TODO: Write a clear description here.
#     # Tell the agent WHEN to use this tool and WHAT input it expects.
#     """
#     # TODO: convert the string to a number
#     # TODO: multiply by 83
#     # TODO: return a formatted string
#     pass

# # Test it directly (without the agent):
# print(currency_converter.invoke("100"))   # Expected: 100 USD = 8300 INR

from langchain.tools import tool

@tool
def currency_converter(amount_usd: str) -> str:
    """
    Convert an amount from USD to INR using a fixed exchange rate.

    Use this tool whenever the user asks to convert a USD amount into INR.
    The input should be a string representing the amount in USD (e.g., "100").
    """
    # Convert the string to a number
    usd = float(amount_usd)

    # Fixed conversion rate
    inr = usd * 83

    # Return formatted result
    if usd.is_integer():
        return f"{int(usd)} USD = {int(inr)} INR"
    else:
        return f"{usd} USD = {inr} INR"


# Test it directly (without the agent):
print(currency_converter.invoke("100"))   # Expected: 100 USD = 8300 INR
```

Create a tool called `student_lookup` that returns a student's course from a dictionary — just like the `weather_db` pattern from class.

Requirements:

* Input is a student name; the lookup should be **case-insensitive** (hint: `.lower().strip()`).
* If the student is not found, return a helpful message listing the available names.

```python

# def student_lookup(name: str) -> str:
#     """
#     # TODO: description
#     """
#     # TODO: create student_db dictionary
#     # TODO: look up with .lower().strip()
#     # TODO: handle 'not found' case
#     pass

# # Test:
# print(student_lookup.invoke("Aryan"))
# print(student_lookup.invoke("Unknown Person"))


@tool
def student_lookup(name: str) -> str:
    """
    Look up a student's course from the student database.

    Use this tool when the user asks which course a student is enrolled in.
    The input should be the student's name as a string.
    """
    # Student database
    student_db = {
        "aryan": "AI & ML",
        "naveen": "Computer Science",
        "priya": "Data Science",
        "rahul": "Cyber Security"
    }

    # Case-insensitive lookup
    student_name = name.lower().strip()

    if student_name in student_db:
        return f"{name.strip()} is enrolled in {student_db[student_name]}."
    else:
        available = ", ".join(student_db.keys())
        return (
            f"Student '{name}' not found. "
            f"Available students: {available}."
        )


# Test:
print(student_lookup.invoke("Aryan"))
print(student_lookup.invoke("Unknown Person"))
```

### Exercise B3 — Safe Calculator Upgrade (Medium)

The class `calculator` tool used `eval()` with an error handler. Extend it:

* If the expression contains the word `"import"` or `"__"`, return `"Error: unsafe expression"` **before** calling eval.
* Add support for `math` functions (already done in class — keep it).
* Test it with a safe input and an unsafe input.

```python

# def safe_calculator(expression: str) -> str:
#     """
#     Useful for solving math calculations.
#     Input must be a valid Python math expression.
#     """
#     # TODO: check for unsafe keywords first
#     # TODO: then eval with {'math': math, '__builtins__': {}}
#     pass

# # Tests:
# print(safe_calculator.invoke("math.sqrt(144)"))          # result: 12.0
# print(safe_calculator.invoke("__import__('os')"))         # Error: unsafe expression

import math

@tool
def safe_calculator(expression: str) -> str:
    """
    Useful for solving math calculations.

    Input must be a valid Python math expression.
    Supports math functions (e.g., math.sqrt, math.sin, math.cos).
    Blocks unsafe expressions containing 'import' or '__'.
    """
    # Check for unsafe keywords first
    if "import" in expression or "__" in expression:
        return "Error: unsafe expression"

    try:
        # Safely evaluate the expression
        result = eval(
            expression,
            {"math": math, "__builtins__": {}},
            {}
        )
        return str(result)
    except Exception as e:
        return f"Error: {e}"


# Tests:
print(safe_calculator.invoke("math.sqrt(144)"))      # 12.0
print(safe_calculator.invoke("__import__('os')"))    # Error: unsafe expression
```

### Exercise C1 — Wire It All Together

Build a ReAct agent that uses **your three tools** from Part B (`currency_converter`, `student_lookup`, `safe_calculator`).

Steps:
1. Put your tools in a list.
2. Reuse the ReAct prompt template from class (copied below).
3. Create the agent with `create_react_agent` and wrap it in an `AgentExecutor` with `verbose=True`, `max_iterations=6`, and `handle_parsing_errors=True`.

```python

my_tools = [
    currency_converter,
    student_lookup,
    safe_calculator
]

react_prompt = PromptTemplate.from_template("""
You are a helpful AI assistant. Answer the question using the tools available.

You have access to these tools:
{tools}

Use EXACTLY this format:

Question: the input question
Thought: think about what to do
Action: one of [{tool_names}]
Action Input: input to the action
Observation: result of the action
... (repeat as needed)
Thought: I now know the final answer
Final Answer: the final answer to the question is : <answer>

Begin!

Question: {input}
Thought:{agent_scratchpad}
""")

my_agent = create_react_agent(
    llm=llm,
    tools=my_tools,
    prompt=react_prompt
)

my_executor = AgentExecutor(
    agent=my_agent,
    tools=my_tools,
    verbose=True,
    max_iterations=6,
    handle_parsing_errors=True
)

print("Agent ready!")
```
